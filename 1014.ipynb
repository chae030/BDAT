```python
num = int(input("영어로 출력하고 싶은 동물 번호 선택 : "))

if num == 1 :
    print("선택한 개의 영어는 Dog 입니다.")
elif num == 2 :
    print("선택한 고양이의 영어는 Cat 입니다.")
elif num == 3 :
    print("선택한 소의 영어는 Cow 입니다.")
elif num == 4 :
    print("선택한 닭의 영어는 Chicken 입니다.")
else : 
    print("I Don't Know")
```

```python
kor_n = ["개", "고양이", "소", "닭"]
eng_n = ["Dog", "Cat", "Cow", "Chicken"]
num = 1

print("*** 동물 리스트 ***")
for k in kor_n :
    print(str(num)+". "+k)
    num = num + 1
    
num = int(input("영어로 출력하고 싶은 동물 번호 선택 : "))

num = num -1
if num >= 0 and num <= 3 : 
    print("선택한 "+kor_n[num]+"의 영어는 "+eng_n[num]+" 입니다.")
else : 
    print("I Don't Know")
```

```python
if (num >= 1 and num <= 4) :
    if (num == 1) :
        kor_n = "개"
        eng_n = "Dog"
    
    print("선택하신 "+kor_n+"은 영어로 "+eng_n+"입니다.")
    
else :
    print("I Don't Know")
```

```python
hight = int(input("키(cm) : "))
weight = int(input("몸무게(kg) : "))

bmi = weight / ((hight * 0.01) ** 2)

if bmi < 18.5 :
    print("저체중")
elif bmi < 23 : 
    print("정상")
elif bmi < 25 :
    print("과체중")
elif bmi < 30 :
    print("비만")
elif bmi < 35 :
    print("고도비만")
else :
    print("초고도비만")
```
