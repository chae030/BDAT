```python
# 마지막에 중간 문구 안나오게 하는법~~

while (True) :
    start = int(input("시작값을 입력하세요 : "))
    stop = int(input("종료깂을 입력하세요 : "))
    step = int(input("증감값을 입력하세요 : "))

    if (start == step) :
        print("시작값과 종료값이 같습니다.\n다시 입력하세요.")
        continue
    elif (start < stop) :
        for i in range(start, stop, step) :
            print(i, end = " 야호 ")
        print(stop)
        break
    else :
        for i in range(start, stop, (-1)*step) :
            print(i, end = " 야호 ")
        print(stop)
        break
```

```python
# 15~40 사이의 정수 중 3의 배수 개수와 7의 배수 개수를 출력하는 프로그램을 작성

x = 0 # 3의배수
y = 0 # 7의배수

for i in range(15, 41) :
    if (i % 3 == 0) :
        x += 1
    if (i % 7 == 0) :
        y += 1

print(f'3의 배수는 {x}개, 7의 배수는 {y}개')
```

```python
# 피타고라스 for문

for a in range(1, 21) :
    for b in range(1, 21) :
        for c in range(1, 21) :
            if ((c**2) == (a**2 + b**2)) :
                print(f'두 변 : {a}, {b}   빗 변 : {c}')
```

```python
# 피타고라스 while문

a = 1
b = 1
c = 1

while (a <= 20) :
    b = 1
    while (b <= 20) :
        c = 1
        while (c <= 20) :
            if ((c**2) == (a**2 + b**2)) :
                print(f'두 변 : {a}, {b}   빗 변 : {c}')
            c += 1
        b += 1
    a += 1
```

```python
# 무한루프 while문

while (True) :
    print("입력값이 모두 0이면 종료합니다.")
    a = int(input("첫 번째 정수를 입력하세요 : "))
    b = int(input("두 번째 정수를 입력하세요 : "))
    
    if ( (a == 0) and (b == 0)) :
        print("모든 수가 0으로 입력되어 종료합니다.")
        break
    print(f'{a} + {b} = {a+b}')
```

```python
# 무한루프 while문 2

while (True) :
    print("입력값이 모두 0이면 종료합니다.")
    a = int(input("첫 번째 정수를 입력하세요 : "))
    b = int(input("두 번째 정수를 입력하세요 : "))
    
    if ( (a == 0) and (b == 0)) :
        print("모든 수가 0으로 입력되어 종료합니다.")
        break
    
    c = input("계산할 연산자를 입력하세요 : ")
    
    if (c == '+') :
        print(f'{a} + {b} = {a+b}')
    elif (c == '-') :
        print(f'{a} - {b} = {a-b}')
    elif (c == '*') :
        print(f'{a} * {b} = {a*b}')
    elif (c == '/') :
        print(f'{a} / {b} = {a/b}')
    elif (c == '//') :
        print(f'{a} // {b} = {a//b}')
    elif (c == '%') :
        print(f'{a} % {b} = {a%b}')
    elif (c == '**') :
        print(f'{a} ** {b} = {a**b}')
    else :
        print("연산자를 잘못 입력하여 계산을 할 수 없습니다.")
```
