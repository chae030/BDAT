## 변수 및 자료형

```python
# 변수 사용 예

# 문제 1)  IoT, 인공지능, 머신러닝에 대한 과목의 총점을 구하고, 평균을 구하시오

iot = 90
ai = 98
ml = 85

sum = iot + ai + ml
mean = sum / 3

print(f'총점은 {sum}점, 평균은 {mean}점')
```

```python
# 변수 실습

# 문제 2) 동네 슈퍼마켓에서 아이스크림 1개, 라면 3봉지, 과자 5봉지 및 빵 3봉지를 골랐다. 총 금액은 얼마인가?
# 단, 변수 5개를 활용하여 작성, 아이스크림 : 1000원, 라면 : 1500원, 과자 : 2000원, 빵 : 1500원

icecream = 1000
ramen = 1500
snack = 2000
bread = 1500

sum = icecream * 1 + ramen * 3 + snack * 5 + bread * 3

print(f'총 금액은 {sum} 원')
```

```python
# 문제 3) 키보드로부터 입력받은 각도(Degree)를 라디안(Radian)으로 변환하여 출력하는 프로그램 작성
# 변환 계산식 - 각도 : 실수로 입력받음, 라디안 : 각도 * (PI/180)

import math

PI = math.pi
degree = int(input("degree : "))
radian = degree * (PI / 180)

print(f"Radian = {radian}")
```

```python
# 문제 4) 키보드로부터 원기둥의 높이와 반지름값을 입력받은 후 원기동의 표면 부피와 표면 면적을 구하여 출력하는 프로그램을 작성하시오.
# 계산식 - 높이 : 실수로 입력받음, 반지름 : 실수로 입력받음, 
# 부피 : PI * 반지름 * 반지름 * 높이, 면적 : ((2 * PI * 반지름) * 높이) + ((PI * 반지름 제곱) * 2)

height = int(input("Height : "))
radius = int(input("Radius : "))

v = PI * (radius ** 2) * height
a = ((2 * PI * radius) * height) + ((PI * (radius ** 2)) * 2)

print(f'Height : {height}, Radius : {radius}, V : {v}, A : {a}')
```

## 조건문

```python
# 실습 문제 1
# 압력 숫자가 짝수인지 홀수인지 계산

num = int(input("숫자를 입력해주세요 : "))

if (num % 2 == 0) :
    print(f'입력하신 숫자인 {num}은 짝수입니다.')
else : 
    print(f'입력하신 숫자인 {num}은 홀수입니다.')
print("프로그램 종료")
```

## 반복문

```python
# 실습 2) 500부터 1000사이에 있는 홀수의 합계 구하는 프로그램

sum = 0

for i in range(500, 1000) :
    if (i % 2) :
        continue
    else :
        sum += i
        
print(sum)
```

```python
# 실습 4) 시작값, 종료값, 증가값을 사용자로부터 입력받고 반복문을 통해 해당 값에 대한 결과값을 출력하는 프로그램을 작성하시오

while (True) :
    start = int(input("시작값 : "))
    stop = int(input("종료값 : "))
    step = int(input("증감값 : "))
    sum = 0
    
    if (start == 0 and stop == 0) : 
        print("프로그램 종료")
        break
    
    for i in range(start, stop + 1, step) :
        sum += i
    
    print(f"{start}부터 {stop}까지 {step}증가하여 더한 합은 : {sum}")
```

```python
# 중첩 for문 실습 2) 문자열 입력받아 모음을 없애고 자음만 출력

string = input("영어 문자열을 입력하세요 : ")

vowel = 'aeiou'
string_consonant = ''

for s in string :
    for v in vowel :
        if (s == v) :
            break
    else:
        string_consonant += s
        
print(string_consonant)
```

```python
# 실습 문제 2) while문을 활용하여 다이아몬드 출력하는 프로그램 작성

star = "\u2605"
count = 9
gap = 4

while (True) :
    print((gap * ' ') + ((9 - gap * 2) * star) + (gap * ' '))
    if (count == 0) :
        break
    elif (count > 5) :
        gap -= 1
    else :
        gap += 1
    count -= 1
```
